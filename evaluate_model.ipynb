```python
# -------------------------- 新增：分批预测工具函数（最小侵入） --------------------------
def batch_predict(model, t, X_discrete, X_continuous, batch_size=512):
    """适配Efin/M3tn等PyTorch模型的分批预测工具，不修改模型内部逻辑"""
    total_samples = X_discrete.shape[0]
    uplift_list = []
    model.eval()  # 保持原eval逻辑
    with torch.no_grad():  # 保持原no_grad逻辑
        for i in range(0, total_samples, batch_size):
            # 分批截取数据（仅修改此处，其余逻辑与原代码一致）
            batch_t = t[i:i+batch_size]
            batch_Xd = X_discrete[i:i+batch_size]
            batch_Xc = X_continuous[i:i+batch_size]
            # 原模型调用逻辑完全保留
            uplift_batch, y_pred_batch, *eps_batch = model(None, batch_t, X_discrete=batch_Xd, X_continuous=batch_Xc)
            # 分批存储结果（避免全量内存占用）
            if uplift_batch is not None:
                uplift_list.append(uplift_batch.detach().cpu().numpy())
    # 合并分批结果（原代码返回格式不变）
    return np.concatenate(uplift_list, axis=0) if uplift_list else None
```
